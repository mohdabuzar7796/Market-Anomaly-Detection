# Run All Notebooks (Sequential)

This runner executes the split notebooks **in order** within the same kernel,
so variables are preserved across sections.

Order:
1. 1.0-abz-preprocessing.ipynb
2. 1.1-abz-eda.ipynb
3. 1.2-abz-features.ipynb
4. 1.3-abz-statistical-modeling.ipynb
5. 1.4-abz-ml-modeling.ipynb
6. 1.5-abz-ensemble-analysis.ipynb

```python
# Notebook imports
from pathlib import Path
import nbformat
from IPython import get_ipython
```

```python
_cwd = Path().resolve()
PROJECT_ROOT = _cwd.parent if _cwd.name == "notebooks" else _cwd
DATA_DIR = PROJECT_ROOT / "data"
PLOTS_DIR = PROJECT_ROOT / "reports" / "figures"

root = PROJECT_ROOT / "notebooks"
notebooks = [
    root / "1.0-abz-preprocessing.ipynb",
    root / "1.1-abz-eda.ipynb",
    root / "1.2-abz-features.ipynb",
    root / "1.3-abz-statistical-modeling.ipynb",
    root / "1.4-abz-ml-modeling.ipynb",
    root / "1.5-abz-ensemble-analysis.ipynb",
]

ip = get_ipython()
if ip is None:
    raise RuntimeError("This runner must be executed inside a Jupyter kernel.")

for nb_path in notebooks:
    if not nb_path.exists():
        raise FileNotFoundError(f"Missing notebook: {nb_path}")
    print(f"\n=== Running: {nb_path.name} ===")
    nb = nbformat.read(nb_path, as_version=4)
    for cell in nb.cells:
        if cell.get("cell_type") != "code":
            continue
        source = cell.get("source", "")
        if source.strip():
            ip.run_cell(source)

print("\nAll notebooks executed successfully.")
```
